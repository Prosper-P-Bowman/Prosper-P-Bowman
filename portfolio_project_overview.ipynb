# Personal Portfolio Project

This notebook documents the personal portfolio website created for the web development project.

The portfolio presents a designer and developer through a centered, responsive layout with a blue visual system, animated slideshow, navigation menu, projects, skills, experience, articles, testimonial, and contact information.

## Portfolio Sections

- Introduction and professional ambitions
- Education and certifications
- Selected projects with roles, technologies, and outcomes
- Technical and communication skills
- Experience and responsibilities
- Testimonial and professional endorsements
- Articles and communication samples
- Contact information and social profiles

## Design Direction

The interface uses a deep ocean-blue background, cyan highlights, soft radial glows, compact navigation, readable desktop typography, and responsive mobile spacing. The hero slideshow moves from left to right and the featured update popup cycles automatically.

```python
projects = [
    {
        'name': 'Northstar',
        'role': 'Lead designer',
        'technology': 'Figma, React',
        'outcome': '+34% activation'
    },
    {
        'name': 'Field Notes',
        'role': 'Design and build',
        'technology': 'Next.js, CMS',
        'outcome': '2.1k readers'
    },
    {
        'name': 'Goodground',
        'role': 'Creative partner',
        'technology': 'Brand, Webflow',
        'outcome': '12 launch mentions'
    }
]
```

```python
for project in projects:
    print(f"{project['name']} | {project['role']} | {project['technology']} | {project['outcome']}")
```

```python
skills = {
    'Product thinking': ['Research', 'Strategy', 'Prototyping'],
    'Visual design': ['UI systems', 'Art direction', 'Figma'],
    'Making it real': ['HTML/CSS', 'JavaScript', 'React'],
    'Clear communication': ['Writing', 'Presenting', 'Mentoring']
}

for category, abilities in skills.items():
    print(f"{category}: {', '.join(abilities)}")
```

## Files

- `index.html` contains the portfolio structure and content.
- `styles.css` contains the responsive visual design and typography.
- `script.js` controls the menu, slideshow, and cycling popup.
- `website.png` is the visual reference used for the centered composition.
